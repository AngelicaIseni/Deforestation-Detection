```python
import numpy as np
import xarray as xr
import rioxarray
from shapely.geometry import box, Point
from rasterio.enums import Resampling

values = np.arange(20, dtype="float32").reshape(4, 5)

toy = xr.DataArray(
    values,
    dims=("y", "x"),
    coords={
        "y": [8950035.0, 8950025.0, 8950015.0, 8950005.0],   # centri dei pixel, da nord a sud
        "x": [580005.0, 580015.0, 580025.0, 580035.0, 580045.0],
    },
    name="toy",
)
toy = toy.rio.write_crs("EPSG:32720")
toy
```
